```python
import re

def extract_video_id(url):

    patterns = [
        r"(?:v=|\/)([0-9A-Za-z_-]{11}).*",
        r"youtu\.be\/([0-9A-Za-z_-]{11})",
        r"youtube\.com\/shorts\/([0-9A-Za-z_-]{11})"
    ]

    for pattern in patterns:
        match = re.search(pattern, url)

        if match:
            return match.group(1)

    raise ValueError("Invalid YouTube URL.")
```

```python
url = "https://youtu.be/ukzFI9rgwfU?si=qUvEE2j9PVwtGCv8"

video_id = extract_video_id(url)

print(video_id)
```

```python
def get_youtube_transcript(youtube_url):

    video_id = extract_video_id(youtube_url)

    print("Video ID:", video_id)
    print("Getting transcript...")

    api = YouTubeTranscriptApi()

    transcript_data = api.fetch(video_id)

    transcript = " ".join(
        item.text for item in transcript_data
    )

    if not transcript.strip():
        raise ValueError(
            "No transcript was found for this video."
        )

    print("Transcript obtained successfully.")

    return transcript
```

```python
def chunk_text(text, chunk_size=250):

    words = text.split()

    chunks = [
        " ".join(words[i:i + chunk_size])
        for i in range(0, len(words), chunk_size)
    ]

    return chunks
```

```python
def summarize_text(text):

    if not text or not text.strip():
        raise ValueError(
            "Transcript is empty."
        )

    chunks = chunk_text(text, chunk_size=250)

    summaries = []

    print("Number of chunks:", len(chunks))

    for i, chunk in enumerate(chunks):

        print(
            f"Summarizing chunk {i + 1}/{len(chunks)}..."
        )

        result = summarizer(
            chunk,
            max_length=120,
            min_length=30,
            do_sample=False
        )

        summaries.append(
            result[0]["summary_text"]
        )

    return " ".join(summaries)
```

```python
def summarize_text(text):

    if not text or not text.strip():
        raise ValueError("Transcript is empty.")

    chunks = chunk_text(text, chunk_size=200)

    summaries = []

    print("Number of chunks:", len(chunks))

    for i, chunk in enumerate(chunks):

        print(f"Summarizing chunk {i + 1}/{len(chunks)}...")

        result = summarizer(
            chunk,
            max_length=100,
            min_length=25,
            do_sample=False
        )

        summaries.append(result[0]["summary_text"])

    combined_summary = " ".join(summaries)

    # Final summarization
    final_result = summarizer(
        combined_summary,
        max_length=180,
        min_length=60,
        do_sample=False
    )

    final_summary = final_result[0]["summary_text"]

    return "📌 Summary\n\n" + final_summary
```

```python
def process_video(youtube_url):

    try:

        print("Step 1: Getting YouTube transcript...")

        transcript = get_youtube_transcript(
            youtube_url
        )

        print("Step 2: Generating AI summary...")

        summary = summarize_text(
            transcript
        )

        print("Process completed successfully.")

        return summary

    except Exception as e:

        error_message = (
            f"ERROR: {type(e).__name__}: {str(e)}"
        )

        print(error_message)

        return error_message
```

```python
with gr.Blocks(
    title="Generative AI Video Summarizer"
) as demo:

    gr.Markdown(
        "# 🤖 Generative AI Video Summarizer"
    )

    gr.Markdown(
        "Paste a YouTube link to generate a summarized transcript."
    )

    youtube_url = gr.Textbox(
        label="YouTube Video URL",
        placeholder="https://www.youtube.com/watch?v=..."
    )

    summarize_button = gr.Button(
        "Summarize Video",
        variant="primary"
    )

    output_text = gr.Textbox(
        label="Summarized Key Points",
        lines=12
    )

    summarize_button.click(
        fn=process_video,
        inputs=youtube_url,
        outputs=output_text
    )
```
